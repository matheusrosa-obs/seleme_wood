## Cálculo Preço

```python
import pandas as pd
import numpy as np

# ==============================================================================
# 1. LIMPEZA DE PREÇO (price_raw -> price_final)
# ==============================================================================
# Regex: Captura números (com ou sem ponto de milhar e vírgula decimal)
df2['price_clean_str'] = df2['price_raw'].astype(str).str.extract(r'(\d+(?:\.\d{3})*(?:,\d+)?)')[0]

# Remove ponto de milhar e troca vírgula por ponto
df2['price_clean_str'] = df2['price_clean_str'].str.replace('.', '', regex=False)
df2['price_clean_str'] = df2['price_clean_str'].str.replace(',', '.', regex=False)

# Converte para float
df2['price_final'] = pd.to_numeric(df2['price_clean_str'], errors='coerce')

# ==============================================================================
# 2. CÁLCULO DE VOLUME E PREÇO POR M3
# ==============================================================================
# Garante que as colunas de dimensão sejam numéricas
cols_dim = ['length_mm', 'width_mm', 'thickness_mm']
for col in cols_dim:
    if col in df2.columns:
        df2[col] = pd.to_numeric(df2[col], errors='coerce')

# ------------------------------------------------------------------------------
# ATUALIZAÇÃO: Recalcula a coluna volume_m3 explicitamente
# ------------------------------------------------------------------------------
# Fórmula: (mm * mm * mm) / 1.000.000.000 = m3
df2['volume_m3'] = (df2['length_mm'] * df2['width_mm'] * df2['thickness_mm']) / 1e9

# Calcula (ou recalcula) o preço/m3 usando a coluna de volume atualizada
# Isso garante que os ajustes manuais nas dimensões reflitam no preço unitário
df2['price_per_m3'] = df2['price_final'] / df2['volume_m3']

# Trata divisão por zero (infinito) e valores nulos
df2['price_per_m3'] = df2['price_per_m3'].replace([np.inf, -np.inf], np.nan)

# Arredondar para 2 casas decimais
df2['price_per_m3'] = df2['price_per_m3'].round(2)

# Limpeza de colunas temporárias (mantendo volume_m3)
df2 = df2.drop(columns=['price_clean_str'], errors='ignore')

# Visualizar resultado com as colunas relevantes atualizadas
cols_view = ['length_mm', 'width_mm', 'thickness_mm', 'volume_m3', 'price_final', 'price_per_m3']
print(df2[cols_view].head())
```

```python
df2.groupby('nm_empresa')['price_per_m3'].mean().round(2).sort_values(ascending=False).plot(kind='bar', title='Preço Médio por m3 por Empresa', figsize=(12, 6))
```
